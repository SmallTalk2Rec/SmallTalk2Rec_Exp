# 동연님 전처리 파일

```python
import pandas as pd

data = pd.read_csv('../../../files/movie_list.csv')
```

### null 제거

```python
df = data.dropna(subset=["영화 감독"])
```

### column 이름 변경

```python
df = df.rename(
    columns={
        "영화제목": "title", 
        "영화 감독": "director", 
        "각본(|| parser)": "screenwriter",
        "왓챠 줄거리": "plot", 
        "왓챠피디아 평점": "rating", 
        "왓챠피디아 평점 수 (명)": "rating_count",
        "주연 (|| parser)": "actors",
        "코멘트 10개 평점 (|| parser)": "comment_ratings",
        "코멘트 10개 멘트 (|| parser)": "comment_texts",
        "장르": "genres",
        "제작 국가(|| parser)": "countries",
        "네이버 누적 관객 수(만명)": "audience",
        "네이버 러닝 타임(분)": "running_time",
        "네이버 19세 여부 (1 or 0)": "adult",
    }
)
```

### 필요없는 컬럼 삭제 - 관객수
### 관객수 단위 설정 맞추기 필요

```python
df = df.drop(columns=['audience'])
```

### object column fillna

```python
object_columns = df.select_dtypes(include="object").columns
```

```python
df[object_columns] = df[object_columns].fillna("")
```

```python
df.head()
```

### || 구분자로 되어 있는 것들 split

```python
for col in ["director", "screenwriter", "actors", "comment_ratings", "genres", "countries", "comment_texts"]:
    df[col] = df[col].apply(lambda x: x.replace('||', '/'))
    df[col] = df[col].apply(lambda x: x.split('/'))

    if col is not 'genres':
        df[col] = df[col].apply(lambda x: ''.join(x).split('||')) # 요것만 바꿈
```

```python
df.head()
```

```python
df.to_csv("../../../files/preprocessed_movie_sample.csv", index=False)
```
